## 1.1 Definition of SDGs

The Sustainable Development Goals (SDGs) were adopted by the United Nations (UN) in the year 2015 to engage in the activities such as ending poverty, protecting the planet, and ensuring that by 2030 all people enjoy peace and prosperity.

There are 17 SDGs integrated which emphasize the balance of social, economic and environmental sustainability. Many developed countries are dedicated to helping those who are the furthest behind in achieving SDG goals. It requires people to collect the creativity, technology and financial resources from all of society to construct SDG development.

Below, we present the 17 SDGs as agreed and adopted by the UN.


**SDG 1. No Poverty**   
Understand the spending patterns on mobile phone services as a proxy indicator of income levels

**SDG 2. Zero Hunger**   
Monitor food prices listed online or crowdsourcing to track food security changes

**SDG 3. Good Health and Well-Being**   
Predict the spread of infectious diseases by mapping the movement of anonymized mobile phone users

**SDG 4. Quality Education**   
Curriculum and course recommendation optimized for an individual's learning need

**SDG 5. Gender Equality**   
Analysis of financial transaction to cluster spending patterns and different levels of economic resilience based on a gender

**SDG 6. Clean Water and Sanitation**   
Internet of Things (IoT) sensors connected to water pumps to track the hygiene level of an individual well

**SDG 7. Affordable and Clean Energy**   
Smart metering apporach to increase or restrict the flow of electricity, gas, or water, which could reduce ensure adequate supply for all at peak periods

**SDG 8. Decent Work and Economic Growth**   
Analysis with global postal traffic to measure the economic growth, remittances, trade, and GDP

**SDG 9. Industry, Innovation and Infrastructure**   
Locational data generated by mobile devices to optimize public transportation routes

**SDG 10. Reduced Inequality**   
Speech-to-text analytics on a local radio content to identify concerns and policy responses from the public

**SDG 11. Sustainable Cities and Communities**   
Remote sensing with satellite imagery to determine whether sufficient green spaces are available per population concentration

**SDG 12. Responsible Consumption and Production**   
Analytics on the online purchase data to identify the pace of transition from non-energy efficient to energy-efficient products.

**SDG 13. Climate Action**    
Deforestation tracker based on combined satellite imagery, crowd-sourced witness accounts, and open data

**SDG 14. Life Below Water**   
Location tracker for the maritime vessels to detect illegal or unregulated fishing activities

**SDG 15. Life on Land**   
Social media monitoring to support disaster management with real-time information on victim location, effects and strength of a forest fire

**SDG 16: Peace, Justice and Strong Institutions**   
Sentiment analysis of social media to reveal public opinion on effective governance, public service delivery or human rights.

**SDG 17: Partnerships to Achieve the Goal**   
Private-Public partnership to enable integration of national statistics along with mobile and internet data for synergy.

## 2.2 Accessing World-Bank Open Data

The World Bank provides several APIs to access various types of data and database. However, the most used is the [World Bank Data API](https://datahelpdesk.worldbank.org/knowledgebase/articles/889386-developer-information-overview). World Bank APIs provide access to various types of data and database such as:

* **Indicators API:**  This API gives programmatic access to time series development data and metadata. It is the most commonly used API.

*  **Data Catalog API:** This API provides information about the thousands of development-relevant datasets available through the World Bank Data Catalog.

*  **Projects API:** This API allows access to World Bank operations data. This includes active, pipeline and closed projects implemented in countries and around the world.

* **Finances API:** This is the API that provides programmatic access to World Bank financial data (loans, credits, financial statements, etc) delivered on the World Bank Finances platform.

*  **Climate Data API:** This API allows access to historical and modelled climate data from the Climate Knowledge Portal.

## 2.3 Data Commons

[Data Commons](https://datacommons.org/), powered by Google, aggregates public data from diverse sources like government stats and global organizations, facilitating easy access and usability to address societal challenges.

It synthesizes a single huge knowledge-graph linking references across datasets, making data about specific entities accessible without needing data cleaning.

It is available to all and it offers tools for various user groups and APIs for developers, aiming to aid data-driven insights in education, research, and decision-making.

Data Commons provides an open knowledge repository that combines data from public datasets into one large knowledge-graph. For example, here are some statements contained in the graph:

* [Santa Clara County](https://browser.datacommons.org/kg?dcid=geoId/06085) is contained in the [State of California](https://browser.datacommons.org/kg?dcid=geoId/06)
* The latitude of [Berkeley, CA](https://browser.datacommons.org/kg?dcid=geoId/0606000) is 37.8703
* [The population of Maryland](https://browser.datacommons.org/kg?dcid=dc/p/psjx4xy30nws1) was [6,003,435 in 2018](https://browser.datacommons.org/kg?dcid=dc%2Fo%2Fx1tlfg4ll9yr9).

In the graph, [*entities*](https://en.wikipedia.org/wiki/Entity) like [Santa Clara County](https://browser.datacommons.org/kg?dcid=geoId/06085) are represented by nodes. Every node has a type corresponding to what the node represents. For example, [California](https://browser.datacommons.org/kg?dcid=geoId/06) is a [State](https://schema.org/State). *Relations* between entities are represented by edges between these nodes. For example, the statement "Santa Clara County is contained in the State of California" is represented in the graph as two nodes: "Santa Clara County" and "California" with an edge labeled "[containedInPlace](https://schema.org/containedInPlace)" pointing from Santa Clara to California. Data Commons closely follows the [Schema.org data model](https://schema.org/docs/datamodel.html) and leverages Schema.org schema to provide a common set of types and properties.

* Goal
* Indicator
* Series
* Target
* User

Each category provides specific data related to its name. For instance, the Goal category provides data related to SDGs, while the Indicator category provides detailed data for each goal's indicators.

### **Fetching Data**




**1. Get a list of all SDG Goals.**

### 3.4.1 Geospatial Interactive Visualization

Let's plot the health facilities on a map using latitude and longitude and indicate on the map the different owners of the facilities

````python
# install and import folium to create interactive maps,
!pip install folium

import folium
````

````python
def get_color(ownership):
    colors = {
        'CHAG': 'blue',
        'Private': 'green',
        'private': 'lightgreen',
        'Government': 'red',
        'government': 'darkred',
        'Quasi-Government': 'purple',
        'Islamic': 'orange',
        'Muslim': 'darkorange',
        'Maternity Home': 'pink',
        'Clinic': 'lightblue',
        'NGO': 'gray',
        'Mission': 'brown'
            }
    return colors.get(ownership, 'black')  # Default to 'black' if ownership type not found


# Drop rows with any NaN values
facilities_data = facilities.dropna()

# Create a map centered around Ghana
ghana_map = folium.Map(location=[7.9465, -1.0232], zoom_start=6)

# Iterate through the facilities data and add small circle markers
for index, row in facilities_data.iterrows():
    # Customize the popup text
    popup_text = f"Facility: {row['FacilityName']}<br> Ownership: {row['Ownership']}"
    color = get_color(row['Ownership'])

    folium.CircleMarker(
        location=[row['Latitude'], row['Longitude']],
        radius=3, # You can adjust the size of the dot here
        color=color,
        fill=True,
        fill_opacity=1,
        popup=popup_text
    ).add_to(ghana_map)

# Show the map
ghana_map
````
